```python
import sqlite3
from datetime import datetime


conn = sqlite3.connect("ChatStorage.sqlite")
cursor = conn.cursor()


cursor.execute("""
SELECT Z_PK, ZTEXT, ZMESSAGEDATE
FROM ZWAMESSAGE
WHERE ZTEXT IS NOT NULL
ORDER BY ZMESSAGEDATE DESC
LIMIT 20;
""")

rows = cursor.fetchall()


for row in rows:
    z_pk, text, apple_time = row
    unix_time = apple_time + 978307200
    readable_time = datetime.fromtimestamp(unix_time)

    print("ID:", z_pk)
    print("Date:", readable_time)
    print("Message:", text)
    print("-" * 50)

conn.close()
```

```python
import pandas as pd

data = []

for row in rows:
    z_pk, text, apple_time = row
    unix_time = apple_time + 978307200
    readable_time = datetime.fromtimestamp(unix_time)

    data.append([z_pk, readable_time, text])

df = pd.DataFrame(data, columns=["ID", "Date", "Message"])
df
```

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

conn = sqlite3.connect("ChatStorage.sqlite")

query = """
SELECT ZMESSAGEDATE
FROM ZWAMESSAGE
WHERE ZTEXT IS NOT NULL;
"""

df = pd.read_sql_query(query, conn)


df["unix_time"] = df["ZMESSAGEDATE"] + 978307200
df["date"] = df["unix_time"].apply(lambda x: datetime.fromtimestamp(x))


df["day"] = df["date"].dt.date

counts = df["day"].value_counts().sort_index()


plt.figure()
counts.plot()
plt.xlabel("Date")
plt.ylabel("Number of Messages")
plt.title("Message Activity Over Time")
plt.show()
```

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

conn = sqlite3.connect("ChatStorage.sqlite")

query = """
SELECT ZMESSAGEDATE, ZTEXT
FROM ZWAMESSAGE
WHERE ZTEXT IS NOT NULL;
"""

df = pd.read_sql_query(query, conn)

df["unix_time"] = df["ZMESSAGEDATE"] + 978307200

df["date"] = df["unix_time"].apply(
    lambda x: datetime.fromtimestamp(x)
)

df["day"] = df["date"].dt.date
```

```python
daily_counts = df["day"].value_counts().sort_index()

plt.figure(figsize=(8,4))

daily_counts.plot()

plt.title("Daily Message Activity")
plt.xlabel("Date")
plt.ylabel("Number of Messages")

plt.show()
```

```python
df["hour"] = df["date"].dt.hour

hour_counts = df["hour"].value_counts().sort_index()

plt.figure(figsize=(8,4))

hour_counts.plot(kind="bar")

plt.title("Message Activity by Hour")
plt.xlabel("Hour")
plt.ylabel("Number of Messages")

plt.show()
```

```python
plt.figure(figsize=(8,4))

df["message_length"] = df["ZTEXT"].astype(str).apply(len)

plt.hist(df["message_length"], bins=10)

plt.title("Message Length Distribution")
plt.xlabel("Message Length")
plt.ylabel("Frequency")

plt.show()
```

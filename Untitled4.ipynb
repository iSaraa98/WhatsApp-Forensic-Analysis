```python
import sqlite3
from datetime import datetime


conn = sqlite3.connect("ChatStorage.sqlite")
cursor = conn.cursor()


cursor.execute("""
SELECT Z_PK, ZTEXT, ZMESSAGEDATE
FROM ZWAMESSAGE
WHERE ZTEXT IS NOT NULL
ORDER BY ZMESSAGEDATE DESC
LIMIT 20;
""")

rows = cursor.fetchall()


for row in rows:
    z_pk, text, apple_time = row
    unix_time = apple_time + 978307200
    readable_time = datetime.fromtimestamp(unix_time)

    print("ID:", z_pk)
    print("Date:", readable_time)
    print("Message:", text)
    print("-" * 50)

conn.close()
```

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

conn = sqlite3.connect("ChatStorage.sqlite")

query = """
SELECT ZMESSAGEDATE
FROM ZWAMESSAGE
WHERE ZTEXT IS NOT NULL;
"""

df = pd.read_sql_query(query, conn)


df["unix_time"] = df["ZMESSAGEDATE"] + 978307200
df["date"] = df["unix_time"].apply(lambda x: datetime.fromtimestamp(x))


df["day"] = df["date"].dt.date

counts = df["day"].value_counts().sort_index()


plt.figure()
counts.plot()
plt.xlabel("Date")
plt.ylabel("Number of Messages")
plt.title("Message Activity Over Time")
plt.show()
```

```python
query = """
SELECT ZFROMJID, COUNT(*) as count
FROM ZWAMESSAGE
WHERE ZTEXT IS NOT NULL
GROUP BY ZFROMJID;
"""

df = pd.read_sql_query(query, conn)

plt.figure()
df.plot(kind='bar', x='ZFROMJID', y='count')
plt.title("Messages per User")
plt.show()
```

```python
data = []

for row in rows:
    z_pk, text, apple_time = row
    unix_time = apple_time + 978307200
    readable_time = datetime.fromtimestamp(unix_time)
    data.append([z_pk, readable_time, text])

df = pd.DataFrame(data, columns=["ID", "date", "Message"])
```

```python
df['date'] = pd.to_datetime(df['date'])

messages_per_day = df.groupby(df['date'].dt.date).size()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
messages_per_day.plot(marker='o')

plt.title("Message Timeline")
plt.xlabel("Date")
plt.ylabel("Number of Messages")


plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
